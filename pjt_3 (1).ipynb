# Questão 1
### (a)

Para provar que o funcional de avaliação $L_X$ é uma transformação linear, precisamos verificar que ele é fechado para combinação linear.
 Tome $f, g \in \mathcal{F}$ e um escalar $\alpha \in \mathbb{R}$. Daí, usando a definição do funcional, temos:
 $$L_X(f + g) = ((f+g)(x_1), \dots, (f+g)(x_m)) = (f(x_1) + g(x_1), \dots, f(x_m) + g(x_m))$$
 Separando a soma em dois vetores em $\mathbb{R}^m$:$$L_X(f + g) = (f(x_1), \dots, f(x_m)) + (g(x_1), \dots, g(x_m)) = L_X(f) + L_X(g)$$
  Ou seja, $L_X$ é fechado para soma. Agora, avaliamos a multiplicação pelo escalar $\alpha$:$$L_X(\alpha f) = ((\alpha f)(x_1), \dots, (\alpha f)(x_m)) = (\alpha f(x_1), \dots, \alpha f(x_m))$$ Como $\alpha$ multiplica cada entrada do vetor, podemos colocá-lo em evidência:$$L_X(\alpha f) = \alpha (f(x_1), \dots, f(x_m)) = \alpha L_X(f)$$
  Logo, $L_X$ é fechado para produto por escalar. Como as duas condições são satisfeitas, concluímos que $L_X$ é uma transformação linear.

### (b)

Para mostrar que $\mathcal{P}_d$ é um subespaço linear de $\mathcal{F}$, precisamos verificar que ele é fechado para soma e multiplicação por escalar. Sejam $p, q \in \mathcal{P}_d$ e $\alpha \in \mathbb{R}$. Podemos expressá-los genericamente como $p(x) = \sum_{i=0}^d a_i x^i$ e $q(x) = \sum_{i=0}^d b_i x^i$, com $a_i, b_i \in \mathbb{R}$.
 Considere a soma:$$(p + q)(x) = p(x) + q(x) = \sum_{i=0}^d a_i x^i + \sum_{i=0}^d b_i x^i = \sum_{i=0}^d (a_i + b_i) x^i$$Como $(a_i + b_i) \in \mathbb{R}$, o resultado é um polinômio de grau no máximo $d$, logo $(p + q) \in \mathcal{P}_d$.
  Agora considere a multiplicação por escalar:$$(\alpha p)(x) = \alpha p(x) = \alpha \sum_{i=0}^d a_i x^i = \sum_{i=0}^d (\alpha a_i) x^i$$Como $(\alpha a_i) \in \mathbb{R}$, o resultado também é um polinômio de grau no máximo $d$, logo $(\alpha p) \in \mathcal{P}_d$. Portanto, $\mathcal{P}_d$ é de fato um subespaço linear.
 Como $\mathcal{P}_d$ é um subespaço linear contido em  $\mathcal{F}$ e $L_X$ é linear em todo o espaço $\mathcal{F}$, as propriedades $L_X(p+q) = L_X(p) + L_X(q)$ e $L_X(\alpha p) = \alpha L_X(p)$ continuam válidas para quaisquer $p, q \in \mathcal{P}_d$. Logo, a restrição de $L_X$ a $\mathcal{P}_d$ é uma transformação linear.

### (c)

Para mostrar que a família $\{p_1, p_2, \dots, p_{d+1}\}$ é linearmente independente, supomos que uma combinação linear desses polinômios resulta no polinômio nulo. Ou seja, tomamos escalares $c_1, c_2, \dots, c_{d+1} \in \mathbb{R}$ tais que, para todo $x$:$$\sum_{i=1}^{d+1} c_i p_i(x) = 0$$Pela definição dada no enunciado, temos que $p_i(x_k) = \delta_{ik}$. Assim, avaliando essa equação em um ponto arbitrário $x_k \in X$:
 $$\sum_{i=1}^{d+1} c_i p_i(x_k) = c_1 p_1(x_k) + \dots + c_k p_k(x_k) + \dots + c_{d+1} p_{d+1}(x_k) = 0$$
  Como $\delta_{kk} = 1$ e $\delta_{ik} = 0$ para qualquer $i \neq k$, todos os termos da soma se anulam, exceto o k-ésimo termo. A equação se reduz a:$$c_k \cdot 1 = 0 \implies c_k = 0$$
   Concluímos que $c_k = 0$ para todo $k \in \{1, 2, \dots, d+1\}$. Logo, a única solução possível é a trivial, o que prova que os polinômios são linearmente independentes.
    Sabemos que o espaço $\mathcal{P}_d$ possui dimensão $d+1$, uma vez que a base canônica $\{1, x, x^2, \dots, x^d\}$ tem exatamente $d+1$ elementos. Como o conjunto $\{p_1, p_2, \dots, p_{d+1}\}$ possui $d+1$ vetores linearmente independentes dentro de um espaço de dimensão $d+1$, esse conjunto também é uma base de $\mathcal{P}_d$.

### (d)
**Caso $m < d+1$:**
Para os $m$ pontos do próprio conjunto $X$, podemos construir os polinômios interpoladores de Lagrange $p_1, \dots, p_m$. Cada $p_i$ tem grau $m-1$. Como $m-1 < d$, todos eles já estão em $\mathcal{P}_d$. Avaliando o funcional, temos $L_X(p_i) = e_i$, onde $e_i$ é o $i$-ésimo vetor da base canônica do $\mathbb{R}^m$. Como a imagem de $L_X$ contém todos os vetores da base do $\mathbb{R}^m$, a transformação cobre todo o contradomínio. Logo, $\text{rank}(L_X) = m$.

**Caso $m > d+1$:**
Vamos olhar só para os primeiros $d+1$ pontos de $X$. Pelo item (c), os polinômios de Lagrange associados a esses pontos formam uma base para $\mathcal{P}_d$. Aplicando $L_X$ nessa base, obtemos $d+1$ vetores em $\mathbb{R}^m$. Como $p_i(x_j) = \delta_{ij}$ para esses $d+1$ pontos, cada vetor resultante tem o valor $1$ em uma entrada e $0$ nas outras $d$ posições correspondentes. Isso garante que esses $d+1$ vetores são LI. Como a dimensão do domínio $\mathcal{P}_d$ é $d+1$, a imagem tem dimensão $d+1$. Logo, $\text{rank}(L_X) = d+1$.

**Caso $m = d+1$:**
Aqui a lógica é basicamente a mesma do caso $m < d+1$. Como $m = d+1$, construímos os $d+1$ polinômios de Lagrange usando todos os pontos de $X$. Avaliando o funcional, caímos direto em $L_X(p_i) = e_i$, o que gera a base canônica inteira do contradomínio $\mathbb{R}^{d+1}$. Assim, a transformação pega o espaço todo e $\text{rank}(L_X) = d+1$.

Portanto, concluímos que $L_X$ é de posto completo, ou seja:

$$\text{rank}(L_X) = \min\{m, d+1\}$$

### (e)

A matriz da transformação $L_X$ na base canônica de $\mathcal{P}_d$, dada por $\{1, x, x^2, \dots, x^d\}$, é construída avaliando cada elemento da base nos $m$ pontos do conjunto $X$. Ou seja, a $i$-ésima linha da matriz é da forma $(1, x_i, x_i^2, \dots, x_i^d)$. Essa é exatamente a definição da matriz de Vandermonde $V$, que possui dimensões $m \times (d+1)$.

O posto de uma transformação linear é o mesmo da matriz que a representa. Do item (d), $\text{rank}(L_X) = \min\{m, d+1\}$, ou seja, $\text{rank}(V) = \min\{m, d+1\}$ e $V$ é de posto completo. No caso $m=d+1$, $V$ é quadrada de posto completo, ou seja, inversível.

### (f)

O polinômio interpolador de Lagrange $p_1(x)$ associado ao ponto $x_1$ em um conjunto de $d+1$ pontos é dado por:$$p_1(x) = \prod_{j=2}^{d+1} \frac{x - x_j}{x_1 - x_j}$$
 Para calcular $p_1(x)$, fazemos, portanto, $d$ subtrações e $d-1$ multiplicações para o numerador e $d$ subtrações e $d-1$ multiplicações para o denominador, além de $1$ divisão entre eles ao final. Isso resulta em um total de $4d - 1$ operações (ou seja, complexidade $\mathcal{O}(d)$).
 
  Para calcular $p_1(x')$ em outro ponto, podemos aproveitar o cálculo do denominador, que não varia conforme a entrada $x$. Assim, podemos aproveitar essas $2d-1$ operações do denominador para avaliar $p_1$ nesse novo ponto.
  
  Perceba que não é necessário calcular os coeficientes de $p_1$, pois o produtório nos permite avaliar diretamente $p_1$ em $x$, sem precisar expandir a fórmula para obter os coeficientes.

### (g)

Para calcular a matriz de $L_X$ na base canônica de $\mathcal{P}_d$, construímos uma matriz de Vandermonde de dimensão $m \times (d+1)$. Cada linha $i$ é da forma $(1, x_i, x_i^2, \dots, x_i^d)$. Assim, para obter $x_i^2$, precisamos de $1$ operação ($x_i \cdot x_i$); para obter $x_i^3$, mais $1$ operação ($x_i^2 \cdot x_i$), e assim por diante, até $x_i^d$. Desse modo, precisamos de $d-1$ operações por linha. Como há $m$ linhas, isso totaliza $m(d-1)$ operações, ou seja, a complexidade é $\mathcal{O}(md)$.


Na base dos polinômios interpoladores de Lagrange, escolhemos $d+1$ pontos de $X$ para construí-la. Tem-se $p_j(x_i) = \delta_{ij}$ para esses pontos específicos, de modo que as linhas correspondentes a eles já estão determinadas ($1$ na coluna $j$ e $0$ nas demais), não sendo necessário realizar operação alguma. Portanto, para $m \le d+1$, nenhuma operação é necessária.Já para $m > d+1$, além das $d+1$ linhas que já estão determinadas, precisamos avaliar os $d+1$ polinômios nos $m-(d+1)$ demais pontos de $X$. Pelo item (f), isso exige:$$(m-d-1)(d+1)(4d-1) = 4md^2 - 4d^3 - 7d^2 + 3md - 2d - m + 1$$operações. Analisando os termos de mais alta ordem, a complexidade é $\mathcal{O}(md^2)$.

Entretanto, a despeito da forma ingênua apresentada acima, existe uma forma mais eficiente de avaliar os polinômios de Lagrange, chamada forma baricêntrica.

Para tal, definimos inicialmente o polinômio nodal:
$$
\ell(x)=\prod_{k=1}^{d+1}(x-x_k).
$$

Sabemos que a fórmula clássica dos polinômios de Lagrange é
$$
p_j(x)=\prod_{\substack{k=1\\k\neq j}}^{d+1}\frac{x-x_k}{x_j-x_k}.
$$

Com isso, o numerador pode ser escrito como
$$
\frac{\ell(x)}{x-x_j},
$$

pois $\ell(x)$ contém todos os fatores $(x-x_k)$, incluindo apenas um fator extra $(x-x_j)$.

Como o denominador depende apenas dos nós $x_k$, definimos os pesos baricêntricos
$$
w_j=\frac{1}{\displaystyle\prod_{\substack{k=1\\k\neq j}}^{d+1}(x_j-x_k)}.
$$

Assim,
$$
p_j(x)=\ell(x)\frac{w_j}{x-x_j}.
$$

A construção da matriz passa então a ter duas etapas.

Na primeira, calculam-se os pesos $w_j$. Como existem $d+1$ pesos e cada um exige um produtório de tamanho $d$, o custo é
$$
O(d^2).
$$

Essa etapa é feita apenas uma vez.

Na segunda etapa, para cada ponto $x_i$, calcula-se primeiro $\ell(x_i)$, com custo $O(d)$. Depois, cada entrada $p_j(x_i)$ é obtida em tempo $O(1)$, pois $w_j$ já está pré-calculado.

Logo, o custo para preencher uma linha inteira da matriz é $O(d)$, e o custo total da etapa dinâmica é
$$
O(md).
$$

Portanto, o custo total da construção é
$$
O(d^2+md).
$$

Como no problema de mínimos quadrados vale $m \ge d+1$, o termo dominante é $O(md)$. Assim, a construção da matriz $L_X$ na base de Lagrange possui a mesma complexidade assintótica da matriz de Vandermonde, mas com condicionamento numérico significativamente melhor, pois as colunas continuam bem linearmente independentes pelo bloco Identidade na parte de cima. Ao contrário da matriz de Vandermonde, onde as colunas tendem a se tornar quase linearmente dependentes  (principalmente para **d** alto, já que $(0.99)^{20} \approx (0.99)^{21}$ e $(0.01)^{20} \approx (0.01)^{21}$) fazendo os valores singulares colapsarem a 0.

#### Comprovação Empírica:

Vemos que as linhas de Vandermonde e Lagrange Baricêntrico sempre são retas, tanto variando **d** quanto variando **m**, o que faz sentido pois $O(md)$ é linear nesses parâmetros. Outra coisa interessante é que Lagrange Ingênuo tem um formato paraboloide variando **d** e um formato retilíneo variando **m**, o que condiz com a complexidade $O(md^2)$ e $O(md)$.

```julia
using LinearAlgebra
using Statistics
using BenchmarkTools
using Plots

function custom_cond(M, d)
    S = svdvals(M)
    return S[1] / S[d + 1]
end

# ─────────────────────────────────────────────────────────────────────────────
# Vandermonde: base canônica -> complexidade O(md)
# ─────────────────────────────────────────────────────────────────────────────
function vandermonde_matrix(X, d)
    """
    Constrói a matriz de Vandermonde m por (d+1).
    Cada linha i é [1, x_i, x_i², ..., x_i^d].
    """
    m = length(X)
    V = ones(m, d + 1)

    for i in 1:m
        xi = X[i]
        for k in 2:(d + 1)
            V[i, k] = V[i, k - 1] * xi
        end
    end

    return V
end


# ─────────────────────────────────────────────────────────────────────────────
# Lagrange ingênuo -> complexidade O(md²)
# ─────────────────────────────────────────────────────────────────────────────
function lagrange_naive(X, d)
    m = length(X)
    # Pega d+1 valores igualmente espaçados no intervalo de X
    nodes = range(minimum(X), maximum(X), length=d+1)

    L = zeros(m, d + 1)

    # Avalia para todos os m pontos
    for i in 1:m
        xi = X[i]

        for j in 1:d+1
            xj = nodes[j]

            num = 1.0
            den = 1.0

            for k in 1:d+1
                if k != j
                    num *= (xi - nodes[k])
                    den *= (xj - nodes[k])
                end
            end

            L[i, j] = num / den
        end
    end

    return L
end


# ─────────────────────────────────────────────────────────────────────────────
# Lagrange baricêntrico -> complexidade O(d² + md)
# ─────────────────────────────────────────────────────────────────────────────
function lagrange_barycentric(X, d)
    m = length(X)
    # Pega d+1 valores igualmente espaçados no intervalo de X
    nodes = range(minimum(X), maximum(X), length=d+1)

    L = zeros(m, d + 1)

    # pesos baricêntricos
    w = ones(d + 1)

    for j in 1:d+1
        for k in 1:d+1
            if k != j
                w[j] *= (nodes[j] - nodes[k])
            end
        end
    end

    w .= 1.0 ./ w

    # Avalia para todos os m pontos
    for i in 1:m
        xi = X[i]
        
        # Verifica se o ponto atual bate exatamente com algum nó para evitar divisão por zero
        exact_idx = findfirst(x -> x == xi, nodes)
        
        if exact_idx !== nothing
            L[i, exact_idx] = 1.0
        else
            ell = prod(xi .- nodes)

            for j in 1:d+1
                L[i, j] = ell * w[j] / (xi - nodes[j])
            end
        end
    end

    return L
end

# ─────────────────────────────────────────────────────────────────────────────
# Função auxiliar de temporização
# ─────────────────────────────────────────────────────────────────────────────
function timeit(fn, args...; reps=10)

    times = Float64[]

    for _ in 1:reps
        t = @elapsed fn(args...)
        push!(times, 1000*t)   # ms
    end

    return minimum(times)
end


# ─────────────────────────────────────────────────────────────────────────────
# Gráfico 1 — variando d (m fixo)
# ─────────────────────────────────────────────────────────────────────────────
m_fixed = 200
d_range = 3:3:48

t_vand_d  = Float64[]
t_naive_d = Float64[]
t_bary_d  = Float64[]

for d in d_range
    X = collect(range(-1, 1, length=m_fixed))

    push!(t_vand_d,  timeit(vandermonde_matrix, X, d))
    push!(t_naive_d, timeit(lagrange_naive, X, d))
    push!(t_bary_d,  timeit(lagrange_barycentric, X, d))
end

p1 = plot(
    d_range, t_vand_d,
    lw=2,
    marker=:circle,
    label="Vandermonde  O(md)",
    xlabel="grau d",
    ylabel="tempo (ms)",
    title="Variando d (m=$m_fixed)"
)

plot!(p1,
    d_range, t_naive_d,
    lw=2,
    marker=:square,
    label="Lagrange ingênuo  O(md²)"
)

plot!(p1,
    d_range, t_bary_d,
    lw=2,
    marker=:utriangle,
    label="Lagrange baricêntrico  O(d²+md)"
)


# ─────────────────────────────────────────────────────────────────────────────
# Gráfico 2 — variando m (d fixo)
# ─────────────────────────────────────────────────────────────────────────────
d_fixed = 15
m_range = (d_fixed+1):30:500

t_vand_m  = Float64[]
t_naive_m = Float64[]
t_bary_m  = Float64[]

for m in m_range
    X = collect(range(-1, 1, length=m))

    push!(t_vand_m,  timeit(vandermonde_matrix, X, d_fixed))
    push!(t_naive_m, timeit(lagrange_naive, X, d_fixed))
    push!(t_bary_m,  timeit(lagrange_barycentric, X, d_fixed))
end

p2 = plot(
    m_range, t_vand_m,
    lw=2,
    marker=:circle,
    label="Vandermonde  O(md)",
    xlabel="número de pontos m",
    ylabel="tempo (ms)",
    title="Variando m (d=$d_fixed)"
)

plot!(p2,
    m_range, t_naive_m,
    lw=2,
    marker=:square,
    label="Lagrange ingênuo  O(md²)"
)

plot!(p2,
    m_range, t_bary_m,
    lw=2,
    marker=:utriangle,
    label="Lagrange baricêntrico  O(d²+md)"
)


# ─────────────────────────────────────────────────────────────────────────────
# Figura final
# ─────────────────────────────────────────────────────────────────────────────
fig = plot(
    p1, p2,
    layout=(1,2),
    size=(1400,500)
)

display(fig)
```

# Questão 2
### (a)
Tome $q \in \mathcal{P}_d$ qualquer. Daí,


$$R_{X,d}(L_X(q)) = \arg\min_{p \in \mathcal{P}_d} ||L_X(p) - L_X(q)||_2^2$$

É fato que para $p = q \implies ||L_X(p) - L_X(q)||_2^2 = 0$. Para garantir que $R_{X,d}(L_X(q)) = q$, e que $q$ é o único que satisfaz $\arg\min_{p \in \mathcal{P}_d} ||L_X(p) - L_X(q)||_2^2 = 0$, verificamos que $L_X$ é injetora, ou seja, $L_X(p) = L_X(q) \implies p = q$.

Para tal, vamos mostrar que $\dim(\mathcal{N}(L_X)) = 0$. Suponha $p \in \mathcal{N}(L_X)$. Então $L_X(p) = (0, \dots, 0)$, ou seja, $p(x_i) = 0$ para todo $x_i \in X$.

Como $p$ tem grau no máximo $d$, $p$ deve possuir no máximo $d$ raízes. Mas $p(x) = 0$ para $m > d$ diferentes valores de $x$, logo, $p$ só pode ser o polinômio identicamente nulo. Ou seja, $\dim(\mathcal{N}(L_X)) = 0$ e $L_X$ é injetiva, como queríamos mostrar.


### (b)

Primeiramente, aproveitando os resultados da Questão 1, podemos expressar a transformação $L_X$ sobre $p \in \mathcal{P}_d$ na forma matricial como $L_X(p) = Vc$, onde $V$ é a matriz de Vandermonde de dimensões $m \times (d+1)$ dos pontos do conjunto $X$, e $c$ é o vetor com os coeficientes de $p$.

Desse modo, o problema de encontrar o polinômio $p$ equivale a encontrar os coeficientes $c$ que minimizam a expressão:
$$\arg\min_{c \in \mathbb{R}^{d+1}} ||Vc - b||_2^2$$

Note que:
$$||Vc-b||_2^2 = (Vc-b)^T(Vc-b) = c^T V^T V c - c^T V^T b - b^T V c + b^T b = c^T V^T V c - 2b^T V c + b^T b$$

Para encontrar o vetor $c$ que minimiza essa função, calculamos o gradiente em relação a $c$ e igualamos ao vetor nulo:
$$\nabla_c (c^T V^T V c - 2b^T V c + b^T b) = 2V^TVc - 2V^Tb = 0$$

Como $m \ge d+1$ e os pontos de $X$ são distintos, $V$ tem posto completo nas colunas, o que garante que $V^TV$ seja inversível. Isolando $c$:
$$V^TVc = V^Tb \implies c = (V^TV)^{-1}V^Tb$$

Assim, transformação $R_{X,d}$ mapeia o vetor $b$ para os coeficientes do polinômio essencialmente através da multiplicação matricial $(V^TV)^{-1}V^Tb$, logo vale a linearidade:

1. $R_{X,d}(b_1 + b_2) = (V^TV)^{-1}V(b_1 + b_2) = (V^TV)^{-1}Vb_1 + (V^TV)^{-1}Vb_2 = R_{X,d}(b_1) + R_{X,d}(b_2)$
2. $R_{X,d}(\alpha b) = (V^TV)^{-1}V(\alpha b) = \alpha((V^TV)^{-1}Vb) = \alpha R_{X,d}(b)$

Logo, concluímos que $R_{X,d}$ é uma transformação linear.

### (c)

Do item anterior, a matriz de $R_{X,d}$ na base canônica é a própria $(V^TV)^{-1}V^T$, onde $V$ é construída avaliando a base canônica $\{1, x, x^2, \dots, x^d\}$ nos pontos $x \in X$.

### (**d** e **e**) Condicionamento de $L_X$ e $R_{X,d}$

Exploramos como o número de condicionamento das matrizes $L_X$ (Vandermonde quando na base canônica) e $R_{X,d} = L_X^+$ 
se comporta ao variar $m$ e $d$.

Para qualquer matriz $A$ de posto completo, $\kappa(A^+) = \kappa(A)$, 
pois os valores singulares de $A^+$ são os inversos dos de $A$, basta olhar a SVD de $A^+$, que é $A^+ = V \Sigma ^{-1} U^*$.

```julia
# ================= Configurações ================= #
m = 1000
d_vals = 2:2:100

# Arrays para armazenar os erros relativos dos coeficientes ||p - p_true||
err_p_normal = zeros(length(d_vals))
err_p_pinv = zeros(length(d_vals))
err_p_qr = zeros(length(d_vals))

# Arrays para armazenar a norma do resíduo ||L*p - b||
res_normal = zeros(length(d_vals))
res_pinv = zeros(length(d_vals))
res_qr = zeros(length(d_vals))

for (i, d) in enumerate(d_vals)
    X = collect(range(-1.0, 1.0, length=m))
    L = vandermonde_matrix(X, d)
    
    # Criamos um polinômio verdadeiro com coeficientes arbitrários (ex: todos 1)
    p_true = ones(d + 1)
    
    # Geramos o vetor b exato
    b = L * p_true
    
    # Método 1: Equações Normais "Na Mão"
    # p = (L^T L)^-1 L^T b
    p_normal = inv(L' * L) * (L' * b)
    
    # Método 2: Decomposição SVD
    p_pinv = pinv(L) * b
    
    # Método 3: Fatoração QR (O operador \ do Julia usa QR para matrizes retangulares)
    p_qr = L \ b
    
    # 1. Calculando o erro nos coeficientes
    err_p_normal[i] = norm(p_normal - p_true) / norm(p_true)
    err_p_pinv[i]   = norm(p_pinv - p_true) / norm(p_true)
    err_p_qr[i]     = norm(p_qr - p_true) / norm(p_true)
    
    # 2. Calculando a norma do resíduo
    res_normal[i] = norm(L * p_normal - b)
    res_pinv[i]   = norm(L * p_pinv - b)
    res_qr[i]     = norm(L * p_qr - b)
end

# ================= Plotagem ================= #
# Gráfico 1: Erro nos coeficientes
p1 = plot(d_vals, err_p_normal .+ 1e-16, label="Equações Normais", lw=3, color=:red)
plot!(p1, d_vals, err_p_pinv .+ 1e-16, label="SVD (pinv)", lw=2, ls=:dash, color=:blue)
plot!(p1, d_vals, err_p_qr .+ 1e-16, label="QR", lw=2, ls=:dot, color=:green)
plot!(p1, yscale=:log10, title="Erro nos Coeficientes ||p - p_{true}||", 
      xlabel="Grau d", ylabel="Erro Relativo (base Canônica)", legend=:bottomright, grid=true)

# Gráfico 2: Norma do Resíduo
p2 = plot(d_vals, res_normal .+ 1e-16, label="Equações Normais", lw=3, color=:red)
plot!(p2, d_vals, res_pinv .+ 1e-16, label="SVD (pinv)", lw=2, ls=:dash, color=:blue)
plot!(p2, d_vals, res_qr .+ 1e-16, label="QR", lw=2, ls=:dot, color=:green)
plot!(p2, yscale=:log10, title="Norma do Resíduo ||L_X p - b||", 
      xlabel="Grau d", legend=:right, grid=true)

plot(p1, p2, layout=(1, 2), size=(1000, 450), margin=5Plots.mm)
```

#### Comentários sobre os gráficos acima

Os gráficos evidenciam a diferença de estabilidade numérica entre os três algoritmos ao resolver o problema de mínimos quadrados, destacando o impacto do tratamento interno de erros numéricos.

**Equações Normais (Vermelho):** É o método menos estável. Ao calcular $L_X^T L_X$, o condicionamento do sistema original é elevado ao quadrado ($\kappa^2$), amplificando drasticamente a perda de precisão. A partir de $d \approx 20$, o método entra em colapso numérico: erra completamente os coeficientes e falha na minimização do resíduo.

**SVD / pseudo-inversa (Azul):** A curva SVD reflete o uso da função `pinv()`, que calcula a Decomposição em Valores Singulares por trás dos panos. A sua resistência superior ocorre devido ao mecanismo de truncamento (o algoritmo identifica e ignora (zera) os valores singulares muito baixos gerados pelo mau condicionamento da matriz). Ao inverter apenas a parte "segura" da informação, evita-se a divisão por números muito próximos a zero, que explodiriam a solução. Contudo, a partir de $d \approx 40$, a matriz se torna tão singular (efeito dos nós igualmente espaçados em graus altos) que a recuperação dos coeficientes exatos se torna impossível com a precisão da máquina.

**Fatoração QR (Verde):** Utilizada por padrão pelo operador `\` do Julia em matrizes retangulares, a fatoração QR também opera diretamente sobre $L_X$, evitando o efeito $\kappa^2$. O gráfico da direita mostra a sua maior vantagem prática, que mesmo na região onde a instabilidade destrói a precisão dos coeficientes originais (a partir de $d \approx 40$), o método QR consegue manter a norma do resíduo estabilizada e extremamente baixa (na casa de $10^{-13}$). Ele encontra uma projeção que se ajusta perfeitamente aos pontos do vetor $b$, absorvendo o ruído numérico internamente sem degradar a aproximação final.

```julia
# ================= Configurações ================= #
m = 1000
d_vals = 2:2:100

# Arrays para armazenar os erros relativos dos coeficientes ||p - p_true||
err_p_normal = zeros(length(d_vals))
err_p_pinv = zeros(length(d_vals))
err_p_qr = zeros(length(d_vals))

# Arrays para armazenar a norma do resíduo ||L*p - b||
res_normal = zeros(length(d_vals))
res_pinv = zeros(length(d_vals))
res_qr = zeros(length(d_vals))

for (i, d) in enumerate(d_vals)
    X = collect(range(-1.0, 1.0, length=m))
    L = lagrange_barycentric(X, d)
    
    # Criamos um polinômio verdadeiro com coeficientes arbitrários (ex: todos 1)
    p_true = ones(d + 1)
    
    # Geramos o vetor b exato
    b = L * p_true
    
    # Método 1: Equações Normais "Na Mão"
    # p = (L^T L)^-1 L^T b
    p_normal = inv(L' * L) * (L' * b)
    
    # Método 2: Decomposição SVD
    p_pinv = pinv(L) * b
    
    # Método 3: Fatoração QR (O operador \ do Julia usa QR para matrizes retangulares)
    p_qr = L \ b
    
    # 1. Calculando o erro nos coeficientes
    err_p_normal[i] = norm(p_normal - p_true) / norm(p_true)
    err_p_pinv[i]   = norm(p_pinv - p_true) / norm(p_true)
    err_p_qr[i]     = norm(p_qr - p_true) / norm(p_true)
    
    # 2. Calculando a norma do resíduo
    res_normal[i] = norm(L * p_normal - b)
    res_pinv[i]   = norm(L * p_pinv - b)
    res_qr[i]     = norm(L * p_qr - b)
end

# ================= Plotagem ================= #
# Gráfico 1: Erro nos coeficientes
p1 = plot(d_vals, err_p_normal .+ 1e-16, label="Equações Normais", lw=3, color=:red)
plot!(p1, d_vals, err_p_pinv .+ 1e-16, label="SVD (pinv)", lw=2, ls=:dash, color=:blue)
plot!(p1, d_vals, err_p_qr .+ 1e-16, label="QR", lw=2, ls=:dot, color=:green)
plot!(p1, yscale=:log10, title="Erro nos Coeficientes ||p - p_{true}||", 
      xlabel="Grau d", ylabel="Erro Relativo (base de Lagrange)", legend=:bottomright, grid=true)

# Gráfico 2: Norma do Resíduo
p2 = plot(d_vals, res_normal .+ 1e-16, label="Equações Normais", lw=3, color=:red)
plot!(p2, d_vals, res_pinv .+ 1e-16, label="SVD (pinv)", lw=2, ls=:dash, color=:blue)
plot!(p2, d_vals, res_qr .+ 1e-16, label="QR", lw=2, ls=:dot, color=:green)
plot!(p2, yscale=:log10, title="Norma do Resíduo ||L_X p - b||", 
      xlabel="Grau d", legend=:bottomright, grid=true)

plot(p1, p2, layout=(1, 2), size=(1000, 450), margin=5Plots.mm)
```

**Erro nos Coeficientes:** As Equações Normais (vermelho) falham muito antes, como esperado, por elevarem o mau condicionamento ao quadrado. Os métodos SVD e QR (azul e verde) conseguem acompanhar juntos e resistem um pouco mais, mas por volta de $d \approx 55$ sofrem um colapso do nada. Nesses dois métodos, erro relativo trava exatamente em $10^0$ (ou seja, 100% de erro). Isso sinaliza que a matriz ficou tão corrompida numericamente que a decomposição falha.

**Norma do Resíduo:** Este é o maior contraste em relação aos testes com a base canônica. Lá, o método QR conseguia absorver a instabilidade e manter o resíduo muito próximo do limite de precisão da máquina ($10^{-13}$). Aqui, o resíduo explode para todos os três métodos, batendo na casa de $10^{10}$.

Agora a parte mais interessante (na nossa opinião)

**Por que a base de Lagrange quebrou até o QR?** Na base canônica, as entradas da matriz de Vandermonde estão sempre limitadas entre -1 e 1 (o problema numérico é que as colunas ficam quase linearmente dependentes). Já na base de Lagrange com pontos igualmente espaçados, a Constante de Lebesgue e o Fenômeno de Runge fazem com que os valores inseridos na matriz $L_X$ atinjam escalas astronômicas nas bordas do intervalo.

Quando o seu código executa a linha $b = L \cdot p_{true}$, a multiplicação lida com números tão absurdos que o vetor $b$ já nasce completamente contaminado por erros massivos de ponto flutuante. O solver (seja SVD ou QR) tenta minimizar o resíduo de um sistema que, numericamente, já perdeu o sentido. O problema deixou de ser apenas "colunas quase L.D." e passou a ser "números grandes demais para a aritmética do computador manter a precisão".

**Conclusão:** A base de Lagrange não resolve o problema sozinha. Se não houver a troca da amostragem (passando a usar nós de Chebyshev), a magnitude explosiva dos valores internos da matriz torna o sistema irresolvível em graus altos, estragando até mesmo as fatorações ortogonais mais robustas (como a QR).

#### (f) Comparação entre complexidade computacional e condicionamento

- Complexidade Computacional: Sobre a construção de $L_X$ na base canônica, preencher a matriz de Vandermonde calculando as potências custa $O(md)$. Na base de Lagrange, utilizando a forma baricêntrica, o custo assintótico predominante para $m \ge d+1$ também é otimizado para $O(md)$. Já para o cálculo de $R_{X,d}$, encontrar a pseudo-inversa e resolver o sistema linear sobredeterminado (utilizando, por exemplo, Fatoração QR) é a operação mais custosa do algoritmo e exige $O(md^2)$ em ambas as abordagens. Assim, o custo total: ambas as bases exigem o mesmo esforço assintótico final de $O(md^2)$ para computar $R_{X,d}$.

- Condicionamento: Na base canônica, o condicionamento $\kappa(L_X) = \kappa(R_{X,d})$ cresce muito com o aumento do grau $d$. Novamente explicando, os monômios ($x^2, x^3, \dots, x^d$) se achatam perto da origem e sobem acentuadamente nas bordas do domínio $[-1, 1]$, tornando as colunas de $L_X$ **quase** linearmente dependentes. Ou seja, o menor valor singular colapsa a zero, fazendo o erro númerico explodir. Na base de lagrange, a construção da matriz segura as funções em nós específicos ($p_j(x_i) = \delta_{ij}$) que garantem a independência linear das colunas por causa desses pontos bem definidos. No caso quadrado ($m=d+1$), as matrizes tornam-se identidades e o condicionamento é perfeitamente $1$. No sistema retangular ($m > d+1$), o erro não colapsa igual na base canônica e se mantém relativamente estável, sem crescer TANTO.

- Conclusão: A base de Lagrange é a melhor escolha. Como a otimização pela forma baricêntrica iguala o custo computacional global de $L_X$ em $O(md)$ e de $R_{X,d}$ em $O(md^2)$, não existe perda de performance ao não usar base canônica. Além disso, adoção de Lagrange permite construir e resolver o sistema usufruindo da sua alta estabilidade numérica, contornando a explosão de erros de arredondamento da matriz de Vandermonde sem "trade-offs" no que diz respeito ao comportamento assintótico do algoritmo. Porém, como vimos nos últimos comentários dos itens **d** e **e**, usar a base de Lagrange não é melhor por si só quando falamos de resolver Mínimos Quadrados, pois a utilização de pontos uniformemente espaçados faz ela ficar ainda pior que a base Canônica (efeito de Runge). Então daí vem a necessidade de usar pontos de Chebyshev para interpolar polinômios nessa base.

# Questão 3

### (a)

Para todo $q \in \mathcal{P}d$, existe $b = L_X(q) \in \mathbb{R}^m$ tal que $R{X,d}(b) = q$. Isso significa que a transformação $R_{X,d}$ é sobrejetora em relação a $\mathcal{P}_d$, ou seja, a sua imagem é o próprio espaço $\mathcal{P}_d$.

Assim, para encontrar a imagem da transformação composta, basta considerarmos o posto de $L_Y$ atuando sobre todo o domínio $\mathcal{P}_d$. Nesse caso, recaímos na questão 1(d), mas agora avaliando $n$ pontos (do conjunto $Y$) sobre o espaço $\mathcal{P}_d$ de dimensão $d+1$.

Como visto anteriormente, o posto do funcional de avaliação é o mínimo entre o número de pontos de avaliação e a dimensão do espaço de polinômios. Logo, o posto da composta é:
$$\text{rank}(L_Y \circ R_{X,d}) = \text{rank}(L_Y) = \min\{n, d+1\}$$

### (b)

##### Argumento geométrico
Pensando intuitivamente na geometria do problema, a base escolhida para $\mathcal{P}_d$ não afeta o resultado da aplicação composta porque o polinômio atua apenas como uma curva determinada no espaço e intermediária para o ajuste em $X$ e a avaliação em $Y$. Dado um vetor $b \in \mathbb{R}^m$, existe um único polinômio $p \in \mathcal{P}_d$ que melhor se ajusta aos pontos de $X$, como provado anteriormente pela injetividade de  $L_X$ para $m \ge d+1$, o que assegura uma única solução para o problema de mínimos quadrados. Em seguida, $L_Y$ simplesmente avalia essa mesma curva $p$ num novo conjunto de dados. Como a curva geométrica no espaço é determinada de modo único pelas posições dos pontos de $X$ e do vetor $b$, a sua altura nos pontos de $Y$ será estritamente a mesma, independentemente da base que utilizarmos para escrevê-la algebricamente.

##### Resolução matricial

Considere $A_{X,d}$ a matriz de $R_{X,d}$ e $V_Y$ a matriz de $L_Y$ na base canônica de $\mathcal{P}d$. A operação composta $L_Y \circ R{X,d}$ no modo matricial é expressa por $V_Y A_{X,d}$.Considere agora que utilizamos uma nova base $\beta$ para $\mathcal{P}d$ e seja $M$ a matriz de conversão da base canônica para a base $\beta$ (consequentemente, $M^{-1}$ converte da base $\beta$ para a base canônica).Desse modo, $R{X,d}$ deve receber um vetor de $\mathbb{R}^m$ e retornar as coordenadas de um polinômio de $\mathcal{P}d$ na base $\beta$. A matriz dessa operação pode ser expressa por $M A{X,d}$.Por outro lado, $L_Y$ deve receber as coordenadas de um polinômio na base $\beta$ e retornar um vetor de $\mathbb{R}^n$. A matriz dessa operação pode ser expressa por $V_Y M^{-1}$, ou seja, primeiro reconvertemos para a base canônica e depois avaliamos normalmente em $Y$.Desse modo, a aplicação composta dessas operações na base $\beta$ é dada de modo matricial por:$$V_Y M^{-1} M A_{X,d} = V_Y A_{X,d}$$Portanto, a base escolhida para $\mathcal{P}d$ não afeta os resultados da aplicação composta $L_Y \circ R{X,d}$.

### (c)

A composta $L_Y \circ R_{X,d}$ é uma interpolação no sentido de que ela encontra o polinômio que melhor se ajusta a um conjunto de dados e avalia-o num novo conjunto de dados, mas pertencente ao mesmo domínio dos dados originais. Isso ocorre do seguinte modo: primeiro, $R_{X,d}$ obtém o polinômio que melhor se ajusta aos dados de $X$ no sentido de mínimos quadrados. Em seguida, $L_Y$ avalia esse polinômio num conjunto de dados $Y$ que pertence ao mesmo domínio $[-1, 1]$ de $X$, o que caracteriza uma aplicação de interpolação.

#### (d)

```julia
using Printf

# Lista estratégica de combinações (d, m, n) garantindo m >= d+1
combinacoes = [
    (d=5,  m=10,  n=50),   # Cenário Base (Tudo pequeno)
    (d=5,  m=10,  n=500),  # Teste: n -> inf (Aumenta n drasticamente)
    (d=5,  m=100, n=50),   # Teste: m -> inf (Superamostragem em X)
    (d=15, m=20,  n=200),  # Cenário Intermediário
    (d=25, m=30,  n=50),   # Teste: d -> inf (Grau alto, n pequeno)
    (d=25, m=30,  n=500),  # Teste: d, n -> inf (Grau alto e malha densa)
    (d=35, m=100, n=500)   # Cenário Extremo
]

println("-------------------------------------------------------------------------")
@printf("%-4s | %-4s | %-4s | %-12s | %-15s | %-15s\n",
        "d", "m", "n", "Erro Matriz", "Cond. (Lagrange)", "Cond. (Canônica)")
println("-------------------------------------------------------------------------")

for c in combinacoes
    d, m, n = c.d, c.m, c.n

    X = collect(range(-1.0, 1.0, length=m))
    Y = collect(range(-1.0, 1.0, length=n))

    # Base Canônica: usa vandermonde_matrix definida na Q1
    Vx = vandermonde_matrix(X, d)
    Vy = vandermonde_matrix(Y, d)
    C_can = Vy * pinv(Vx)

    # Base de Lagrange baricêntrica: usa lagrange_barycentric definida na Q1.
    # Os nós são os primeiros d+1 pontos de X (mesma convenção da Q1).
    Lx = lagrange_barycentric(X, d)
    Ly = lagrange_barycentric(Y, d)
    C_lag = Ly * pinv(Lx)

    # Valores Singulares
    sv_can = svdvals(C_can)
    sv_lag = svdvals(C_lag)

    # Condicionamento da Composta (sigma_1 / sigma_{d+1})
    cond_lag = sv_lag[1] / sv_lag[d+1]

    # A base canônica colapsa tão feio que sigma_{d+1} pode dar zero/ruído.
    # O max() garante que não dê erro de divisão por zero na impressão.
    cond_can = sv_can[1] / max(sv_can[d+1])

    # Erro Máximo entre as duas representações da mesma composta
    erro_matriz = norm(C_can - C_lag, Inf)

    @printf("%-4d | %-4d | %-4d | %-12.2e | %-15.2e | %-15.2e\n",
            d, m, n, erro_matriz, cond_lag, cond_can)
end
println("-------------------------------------------------------------------------")
```

A tabela compara os dois caminhos de cálculo da mesma transformação composta $L_Y \circ R_{X,d}$ — pela base canônica e pela base de Lagrange — e revela três fatos centrais:

**Erro entre matrizes:** Os valores da coluna "Erro Matriz" mostram que as duas representações concordam muito bem para graus baixos (erro $\approx 10^{-12}$), mas divergem significativamente para graus altos ($d = 35$). Isso **não** significa que a transformação composta seja diferente — ela é a mesma, como provamos no item (b). O que muda é o erro numérico acumulado durante os cálculos matriciais: a base canônica, com seu condicionamento exponencialmente pior, amplifica os erros de ponto flutuante de forma muito mais intensa.

**Condicionamento:** O condicionamento da composta na base de Lagrange cresce muito mais lentamente com $d$ do que na base canônica. Mais importante ainda, ao aumentar $n$ mantendo $d$ fixo, o condicionamento permanece essencialmente constante — isso é exatamente o item (e) que analisaremos a seguir.

**Independência de base:** A igualdade das matrizes (para $d$ pequeno) confirma numericamente o resultado teórico do item (b): a composta $L_Y \circ R_{X,d}$ é invariante à escolha de base em $\mathcal{P}_d$.

### (e) Condicionamento da composta; $n \to \infty$ e $d \to \infty$

**Condicionamento teórico.** Pela definição, o condicionamento de uma aplicação linear é a razão entre o maior e o menor valor singular não nulo. A composta $C = L_Y \circ R_{X,d}$ mapeia $\mathbb{R}^m \to \mathbb{R}^n$ com posto $r = \min\{n, d+1\}$, de modo que

$$\kappa(C) = \frac{\sigma_1(C)}{\sigma_r(C)}.$$

Intuitivamente, $R_{X,d}$ é a pseudo-inversa de $L_X$, e $L_Y$ é uma avaliação sobre o mesmo espaço $\mathcal{P}_d$. Ambas têm condicionamentos que dependem de $d$ (e na base canônica, crescem exponencialmente com $d$). O condicionamento da composta herda, portanto, esse comportamento com $d$.

**Podemos fazer $n \to \infty$ sem prejudicar o condicionamento?** Sim! O aumento de $n$ introduz mais linhas em $L_Y$, mas o posto da composta fica limitado a $\min\{n, d+1\} = d+1$ para $n > d$. Os $d+1$ valores singulares relevantes dependem apenas de como os polinômios de grau $\leq d$ se comportam nos pontos de $Y$, e essa geometria não muda qualitativamente com $n$ (desde que $Y$ seja uma malha razoavelmente densa em $[-1,1]$). Logo, $\kappa(C)$ estabiliza conforme $n \to \infty$ com $d$ fixo.

**E $n \to \infty$ e $d \to \infty$ simultaneamente?** Nesse caso, o condicionamento **explode**. O problema é inteiramente dominado por $d$: para pontos igualmente espaçados, os polinômios de grau alto oscilar violentamente (Fenômeno de Runge), fazendo os valores singulares pequenos colapsar exponencialmente a zero. Aumentar $n$ não salva esse comportamento.

Verificamos esses fatos experimentalmente abaixo:


```julia
# Teste 1: condicionamento da composta variando n (d e m fixos)
d_fixed = 10
m_fixed = 50
n_vals  = (d_fixed+1):10:2000

cond_n_can = Float64[]
cond_n_lag = Float64[]

X = collect(range(-1.0, 1.0, length=m_fixed))
Vx = vandermonde_matrix(X, d_fixed)
Lx = lagrange_barycentric(X, d_fixed)

for n in n_vals
    Y  = collect(range(-1.0, 1.0, length=n))

    Vy = vandermonde_matrix(Y, d_fixed)
    Ly = lagrange_barycentric(Y, d_fixed)

    C_can = Vy * pinv(Vx)
    C_lag = Ly * pinv(Lx)

    push!(cond_n_can, custom_cond(C_can, d_fixed))
    push!(cond_n_lag, custom_cond(C_lag, d_fixed))
end

p1 = plot(collect(n_vals), cond_n_can,
    label="Canônica", lw=2, color=:red,
    xlabel="n", ylabel="κ(C)", yscale=:log10,
    title="Condicionamento vs n (d=$d_fixed, m=$m_fixed)",
    legend=:topright)
plot!(p1, collect(n_vals), cond_n_lag,
    label="Lagrange", lw=2, color=:blue, ls=:dash)

# Teste 2: condicionamento da composta variando d (m=2d+1, n=500 fixo)
d_vals2 = 2:2:100
n_fixed  = 500

cond_d_can = Float64[]
cond_d_lag = Float64[]

for d in d_vals2
    m = 2d + 1
    X = collect(range(-1.0, 1.0, length=m))
    Y = collect(range(-1.0, 1.0, length=n_fixed))

    Vx = vandermonde_matrix(X, d)
    Vy = vandermonde_matrix(Y, d)
    Lx = lagrange_barycentric(X, d)
    Ly = lagrange_barycentric(Y, d)

    C_can = Vy * pinv(Vx)
    C_lag = Ly * pinv(Lx)

    push!(cond_d_can, custom_cond(C_can, d))
    push!(cond_d_lag, custom_cond(C_lag, d))
end

p2 = plot(collect(d_vals2), cond_d_can,
    label="Canônica", lw=2, color=:red,
    xlabel="d", ylabel="κ(C)", yscale=:log10,
    title="Condicionamento vs d (n=$n_fixed, m=2d+1)",
    legend=:topleft)
plot!(p2, collect(d_vals2), cond_d_lag,
    label="Lagrange", lw=2, color=:blue, ls=:dash)

plot(p1, p2, layout=(1,2), size=(1000, 450), margin=5Plots.mm)
```

#### Comentários sobre os gráficos acima

**Gráfico da esquerda ($n$ variando):** Exatamente como previsto, o condicionamento da composta se estabiliza rapidamente após $n \approx d+1$ em ambas as bases. Isso ocorre porque, para $n > d+1$, o posto da matriz composta já atingiu o seu máximo ($d+1$), e os valores singulares relevantes não se alteram com o acréscimo de mais linhas em $L_Y$. A chave é que $Y \subset [-1,1]$: adicionamos pontos de avaliação dentro do domínio de ajuste, sem extrapolação, e o problema permanece bem-posto.

**Gráfico da direita ($d$ variando):** O condicionamento cresce exponencialmente com $d$, confirmando que $n \to \infty$ não basta — é o crescimento do grau que determina a instabilidade. Na base canônica, a saturação no limite de Float64 aparece por volta de $d \approx 30$; na base de Lagrange, o crescimento é mais lento mas igualmente exponencial com pontos igualmente espaçados.


### (f) Pontos de Chebyshev em $X$

Substituímos agora os pontos igualmente espaçados de $X$ pelos **pontos de Chebyshev**:
$$x_i = \cos\!\left(\frac{2i-1}{2m}\,\pi\right), \quad i = 1, \ldots, m.$$

Esses pontos se concentram nas bordas de $[-1,1]$, onde o Fenômeno de Runge é mais severo. A teoria de aproximação polinomial garante que, com pontos de Chebyshev, a constante de Lebesgue cresce apenas como $O(\log d)$ — em vez de exponencialmente como nos pontos igualmente espaçados — tornando a interpolação muito mais estável.

Para usar as funções já definidas (que constroem a base de Lagrange a partir dos pontos de $X$), basta passar os pontos de Chebyshev como entrada. A função `lagrange_barycentric` já aceita qualquer conjunto de nós.


```julia
# Função auxiliar: pontos de Chebyshev em m pontos
chebyshev_nodes(m) = [cos((2i - 1) / (2m) * π) for i in 1:m]

# ─────────────────────────────────────────────────────────────────
# Comparação dos valores singulares: igualmente espaçados vs Chebyshev
# ─────────────────────────────────────────────────────────────────
d_test = 20
m_test = 50
n_test = 200

X_uniform   = collect(range(-1.0, 1.0, length=m_test))
X_cheby     = chebyshev_nodes(m_test)
Y           = collect(range(-1.0, 1.0, length=n_test))

# Composta com X uniforme (base Lagrange)
Lx_u  = lagrange_barycentric(X_uniform, d_test)
Ly    = lagrange_barycentric(Y, d_test)       # Y não muda
C_u   = Ly * pinv(Lx_u)

# Composta com X Chebyshev (base Lagrange)
Lx_c  = lagrange_barycentric(X_cheby,   d_test)
C_c   = Ly * pinv(Lx_c)

sv_u = svdvals(C_u)
sv_c = svdvals(C_c)

p1 = plot(sv_u[1:d_test+1], marker=:circle, lw=2, color=:orange,
    label="X uniforme", yscale=:log10,
    xlabel="índice", ylabel="σᵢ",
    title="Valores singulares da composta\nd=$d_test, m=$m_test, n=$n_test")
plot!(p1, sv_c[1:d_test+1], marker=:diamond, lw=2, color=:teal,
    label="X Chebyshev")

# ─────────────────────────────────────────────────────────────────
# Condicionamento vs d: uniforme vs Chebyshev (mesmos m, n)
# ─────────────────────────────────────────────────────────────────
d_vals3 = 2:2:40
m3      = 100
n3      = 500

cond_u = Float64[]
cond_c = Float64[]

for d in d_vals3
    Xu  = collect(range(-1.0, 1.0, length=m3))
    Xc  = chebyshev_nodes(m3)
    Y3  = collect(range(-1.0, 1.0, length=n3))

    Ly3  = lagrange_barycentric(Y3, d)

    Cu   = Ly3 * pinv(lagrange_barycentric(Xu, d))
    Cc   = Ly3 * pinv(lagrange_barycentric(Xc, d))

    push!(cond_u, custom_cond(Cu, d))
    push!(cond_c, custom_cond(Cc, d))
end

p2 = plot(collect(d_vals3), cond_u, lw=2, color=:orange,
    marker=:circle, label="X uniforme",
    xlabel="d", ylabel="κ(C)", yscale=:log10,
    title="Condicionamento vs d\nm=$m3, n=$n3",
    legend=:topleft)
plot!(p2, collect(d_vals3), cond_c, lw=2, color=:teal,
    marker=:diamond, label="X Chebyshev")

plot(p1, p2, layout=(1,2), size=(1000, 450), margin=5Plots.mm)
```

#### Comentários sobre os gráficos acima

**Valores singulares (esquerda):** Com $X$ uniforme, os valores singulares menores colapsam rapidamente a zero (ou ao ruído numérico), indicando que a composta está numericamente "esquecendo" direções do espaço de polinômios. Com pontos de Chebyshev, o espectro é muito mais plano: os $d+1$ valores singulares relevantes ficam comparáveis em magnitude, o que é a assinatura de uma transformação bem condicionada.

**Condicionamento vs $d$ (direita):** A diferença é dramática. Com $X$ uniforme, o condicionamento explode exponencialmente já para $d \approx 20$. Com pontos de Chebyshev, o crescimento é muito mais lento e, para graus moderados, a composta permanece computacionalmente utilizável. Isso explica por que os pontos de Chebyshev são o padrão em aproximação polinomial numérica: eles minimizam a constante de Lebesgue, que controla diretamente o condicionamento do problema de interpolação.


### (g) Extrapolação 1: $X = \text{linspace}(-1, 1, m)$, $Y = \text{linspace}(-1.5, 1.5, n)$

Agora $Y$ estende além do intervalo de ajuste $[-1, 1]$. Isso caracteriza **extrapolação**: o polinômio é ajustado em $X$ e avaliado fora desse domínio.

**O que esperar teoricamente?** Polinômios são globalmente definidos, então a operação matemática continua válida. Porém, o condicionamento piora significativamente fora do intervalo de ajuste. Intuitivamente, pequenos erros nos coeficientes do polinômio — amplificados pelo mau condicionamento de Vandermonde — se traduzem em erros grandes nas avaliações extrapoladas, pois os polinômios de grau alto oscilam violentamente fora do domínio de ajuste (crescem sem controle).


```julia
# Comparação interpolação vs extrapolação: valores singulares e condicionamento
d_extrap = 10
m_extrap = 50
n_extrap = 200

X_e = collect(range(-1.0,  1.0, length=m_extrap))
Y_interp = collect(range(-1.0,  1.0, length=n_extrap))   # interpolação
Y_extrap = collect(range(-1.5,  1.5, length=n_extrap))   # extrapolação

Lx_e = lagrange_barycentric(X_e, d_extrap)
Pinv_Lx = pinv(Lx_e)

Ly_i  = lagrange_barycentric(Y_interp, d_extrap)
Ly_ex = lagrange_barycentric(Y_extrap, d_extrap)

C_interp = Ly_i  * Pinv_Lx
C_extrap = Ly_ex * Pinv_Lx

sv_i  = svdvals(C_interp)
sv_ex = svdvals(C_extrap)

p1 = plot(sv_i[1:d_extrap+1],  marker=:circle,  lw=2, color=:blue,
    label="Y ⊆ [-1,1] (interp.)", yscale=:log10,
    xlabel="índice", ylabel="σᵢ",
    title="Valores singulares\nd=$d_extrap, m=$m_extrap")
plot!(p1, sv_ex[1:d_extrap+1], marker=:diamond, lw=2, color=:red,
    label="Y = [-1.5,1.5] (extrap.)")

# Condicionamento vs d: interpolação vs extrapolação
d_range4 = 2:2:30
m4       = 50
n4       = 300

cond_i_vals  = Float64[]
cond_ex_vals = Float64[]

for d in d_range4
    Xv  = collect(range(-1.0, 1.0, length=m4))
    Yi  = collect(range(-1.0, 1.0, length=n4))
    Yex = collect(range(-1.5, 1.5, length=n4))

    Lx4 = lagrange_barycentric(Xv, d)
    P4  = pinv(Lx4)

    Ci  = lagrange_barycentric(Yi,  d) * P4
    Cex = lagrange_barycentric(Yex, d) * P4

    push!(cond_i_vals,  custom_cond(Ci,  d))
    push!(cond_ex_vals, custom_cond(Cex, d))
end

p2 = plot(collect(d_range4), cond_i_vals,  lw=2, color=:blue, marker=:circle,
    label="Interpolação",
    xlabel="d", ylabel="κ(C)", yscale=:log10,
    title="Condicionamento vs d\nm=$m4, n=$n4",
    legend=:topleft)
plot!(p2, collect(d_range4), cond_ex_vals, lw=2, color=:red, marker=:diamond,
    label="Extrapolação [-1.5,1.5]")

plot(p1, p2, layout=(1,2), size=(1000, 450), margin=5Plots.mm)
```

#### Comentários sobre os gráficos acima

**Valores singulares (esquerda):** Na interpolação (azul), os valores singulares relevantes se mantêm na mesma ordem de grandeza. Na extrapolação (vermelho), o maior valor singular $\sigma_1$ cresce de forma expressiva enquanto os menores continuam similares, fazendo o condicionamento explodir. Isso reflete que a composta está "amplificando" certas direções do espaço de polinômios ao avaliar fora do domínio de ajuste.

**Condicionamento vs $d$ (direita):** A extrapolação é sempre mais mal condicionada do que a interpolação, e a diferença cresce com $d$. Mesmo para graus moderados ($d \approx 10$), a extrapolação já apresenta condicionamento ordens de grandeza pior. Isso é a versão numérica do fenômeno clássico: polinômios de grau alto são excelentes interpoladores dentro do domínio de ajuste, mas péssimos extrapoladores fora dele.


### (h) Adianta usar os pontos de Chebyshev em $X$ para a Extrapolação 1?

Com $Y = \text{linspace}(-1.5, 1.5, n)$, substituímos $X$ pelos pontos de Chebyshev e investigamos se o condicionamento melhora. 

**Intuição:** Os pontos de Chebyshev minimizam o erro de interpolação *dentro* de $[-1,1]$, pois são os zeros do polinômio de Chebyshev de grau $m$ nesse intervalo. Fora do intervalo, essa propriedade deixa de valer: o polinômio mínimo-max em $[-1,1]$ não tem nenhuma garantia especial em $[-1.5, 1.5]$. Portanto, o ganho deve existir mas ser menor do que no caso de interpolação pura.


```julia
# Extrapolação 1: uniforme vs Chebyshev em X, com Y = [-1.5, 1.5]
d_range5 = 2:2:30
m5       = 50
n5       = 300

cond_u_extrap = Float64[]
cond_c_extrap = Float64[]

for d in d_range5
    Xu  = collect(range(-1.0, 1.0, length=m5))
    Xc  = chebyshev_nodes(m5)
    Yex = collect(range(-1.5, 1.5, length=n5))

    Ly5 = lagrange_barycentric(Yex, d)

    Cu5  = Ly5 * pinv(lagrange_barycentric(Xu, d))
    Cc5  = Ly5 * pinv(lagrange_barycentric(Xc, d))

    push!(cond_u_extrap, custom_cond(Cu5, d))
    push!(cond_c_extrap, custom_cond(Cc5, d))
end

# Pega também os valores de interpolação do item (g) para comparar
cond_i_ref = Float64[]
for d in d_range5
    Xu  = collect(range(-1.0, 1.0, length=m5))
    Yi  = collect(range(-1.0, 1.0, length=n5))
    Ci  = lagrange_barycentric(Yi, d) * pinv(lagrange_barycentric(Xu, d))
    push!(cond_i_ref, custom_cond(Ci, d))
end

p = plot(collect(d_range5), cond_i_ref,    lw=2, color=:blue,   marker=:circle,
    label="Interp. (X uniforme)",
    xlabel="d", ylabel="κ(C)", yscale=:log10,
    title="Extrap. [-1.5,1.5]: uniforme vs Chebyshev\nm=$m5, n=$n5",
    legend=:topleft)
plot!(p, collect(d_range5), cond_u_extrap, lw=2, color=:red,
    label="Extrap. (X uniforme)")
plot!(p, collect(d_range5), cond_c_extrap, lw=2, color=:green,  marker=:diamond,
    label="Extrap. (X Chebyshev)")

display(p)
```

#### Comentários sobre o gráfico acima

A curva verde (Chebyshev + extrapolação) fica consistentemente abaixo da curva vermelha (uniforme + extrapolação) para todo $d$ testado, confirmando que **sim, adianta usar Chebyshev**, mesmo para extrapolação. O ganho é substancial para graus intermediários ($d \approx 10{-}20$), onde a diferença no condicionamento pode chegar a várias ordens de grandeza.

Entretanto, a curva verde ainda fica acima da curva azul (interpolação pura com $X$ uniforme), mostrando que Chebyshev não elimina completamente o mal condicionamento da extrapolação. A concentração dos pontos nas bordas de $[-1,1]$ é benéfica para controlar as oscilações dentro do intervalo, mas a avaliação em $[-1.5, 1.5]$ exige que o polinômio seja preciso fora dessas bordas, onde nenhum ponto de ajuste existe.

**Conclusão:** Chebyshev ajuda, mas não resolve o problema fundamental da extrapolação: avaliar um polinômio de grau alto fora do seu domínio de ajuste é inerentemente instável, independentemente da escolha dos pontos em $X$.


### (i) Extrapolação 2: $X = \text{linspace}(0, 1, m)$, $Y = \text{linspace}(-1, 0, n)$

Neste cenário, os intervalos de ajuste e avaliação são **disjuntos**: ajustamos o polinômio em $[0,1]$ e o avaliamos em $[-1, 0]$. Isso é extrapolação "máxima" no sentido de que nenhum ponto de avaliação coincide com o domínio de ajuste — ao contrário do item (g), onde $[-1.5, 1.5]$ continha $[-1,1]$ como subconjunto.

**O que esperar?** O condicionamento deve ser ainda pior do que no caso anterior. Os polinômios ajustados em $[0,1]$ têm seus graus de liberdade completamente determinados por valores nesse intervalo; avaliá-los em $[-1,0]$ equivale a uma extrapolação de amplitude comparável ao intervalo de ajuste, o que leva a oscilações ainda maiores.


```julia
# Extrapolação 2: X em [0,1], Y em [-1,0]
d_range6 = 2:2:30
m6       = 50
n6       = 300

cond_extrap2_u  = Float64[]
cond_extrap2_c  = Float64[]

for d in d_range6
    Xu2  = collect(range(0.0,  1.0, length=m6))
    # Pontos de Chebyshev mapeados para [0,1]: xi = 0.5 + 0.5*cos((2i-1)/2m * π)
    Xc2  = [0.5 + 0.5 * cos((2i-1)/(2m6) * π) for i in 1:m6]
    Y2   = collect(range(-1.0, 0.0, length=n6))

    # Lagrange com nós em [0,1]
    Lx_u2 = lagrange_barycentric(Xu2, d)
    Lx_c2 = lagrange_barycentric(Xc2, d)
    Ly2   = lagrange_barycentric(Y2,  d)

    Cu2 = Ly2 * pinv(Lx_u2)
    Cc2 = Ly2 * pinv(Lx_c2)

    push!(cond_extrap2_u, custom_cond(Cu2, d))
    push!(cond_extrap2_c, custom_cond(Cc2, d))
end

# Referência: interpolação padrão em [-1,1]
cond_interp_ref = Float64[]
for d in d_range6
    Xu  = collect(range(-1.0, 1.0, length=m6))
    Yi  = collect(range(-1.0, 1.0, length=n6))
    Ci  = lagrange_barycentric(Yi, d) * pinv(lagrange_barycentric(Xu, d))
    push!(cond_interp_ref, custom_cond(Ci, d))
end

p_e2 = plot(collect(d_range6), cond_interp_ref,   lw=2, color=:blue,   marker=:circle,
    label="Interp. padrão [-1,1]",
    xlabel="d", ylabel="κ(C)", yscale=:log10,
    title="Extrap. 2 (X=[0,1], Y=[-1,0])\nm=$m6, n=$n6",
    legend=:topleft)
plot!(p_e2, collect(d_range6), cond_extrap2_u, lw=2, color=:red,
    label="Extrap. 2 (X uniforme)")
plot!(p_e2, collect(d_range6), cond_extrap2_c, lw=2, color=:green,  marker=:diamond,
    label="Extrap. 2 (X Chebyshev em [0,1])")

display(p_e2)
```

```julia
# Comparacao direta:
# 1) Interpolacao em [-1,1]
# 2) Extrapolacao: ajuste em [0,1], avaliacao em [-1,0]
#
# Em cada caso com:
# - pontos uniformemente espacados
# - nos de Chebyshev
#
# Usamos a base Vandermonde, seguindo o mesmo padrao do codigo anterior.

using Plots

# Funcao-teste classica para mostrar instabilidade com nos uniformes
f_test(x) = 1 / (1 + 25x^2)

# Nos de Chebyshev no intervalo [a,b]
function chebyshev_nodes(a, b, n)
    k = 0:n-1
    x = cos.((2k .+ 1) .* pi ./ (2n))
    return sort((a + b)/2 .+ (b - a)/2 .* x)
end

# Parametros
d = 18
n = d + 1

# Malhas de avaliacao
x_interp = collect(range(-1.0, 1.0, length=800))

x_fit_ext = collect(range(0.0, 1.0, length=500))
x_out_ext = collect(range(-1.0, 0.0, length=500))
x_all_ext = collect(range(-1.0, 1.0, length=1000))

plts = []

# ============================================================
# 1) INTERPOLACAO EM [-1,1] COM PONTOS UNIFORMES
# ============================================================

X_uni_int = collect(range(-1.0, 1.0, length=n))
b_uni_int = f_test.(X_uni_int)

V_uni_int = vandermonde_matrix(X_uni_int, d)
coef_uni_int = V_uni_int \ b_uni_int

y_uni_int = vandermonde_matrix(x_interp, d) * coef_uni_int
y_true_int = f_test.(x_interp)

err_uni_int = round(maximum(abs.(y_uni_int - y_true_int)), sigdigits=3)

p1 = plot(
    x_interp, y_true_int,
    lw=1.5, ls=:dot, color=:gray,
    label="f verdadeira",
    xlims=(-1.05, 1.05), ylims=(-0.5, 2.0),
    title="Interpolação: pontos uniformes",
    xlabel="x", ylabel="p(x)",
    legendfontsize=7
)

plot!(
    p1, x_interp, y_uni_int,
    lw=2, color=:blue,
    label="Interpolador, d=$d, erro=$err_uni_int"
)

scatter!(
    p1, X_uni_int, b_uni_int,
    ms=3, color=:black,
    label="pontos de interpolação"
)

push!(plts, p1)


# ============================================================
# 2) INTERPOLACAO EM [-1,1] COM NOS DE CHEBYSHEV
# ============================================================

X_cheb_int = chebyshev_nodes(-1.0, 1.0, n)
b_cheb_int = f_test.(X_cheb_int)

V_cheb_int = vandermonde_matrix(X_cheb_int, d)
coef_cheb_int = V_cheb_int \ b_cheb_int

y_cheb_int = vandermonde_matrix(x_interp, d) * coef_cheb_int

err_cheb_int = round(maximum(abs.(y_cheb_int - y_true_int)), sigdigits=3)

p2 = plot(
    x_interp, y_true_int,
    lw=1.5, ls=:dot, color=:gray,
    label="f verdadeira",
    xlims=(-1.05, 1.05), ylims=(-0.5, 2.0),
    title="Interpolação: nós de Chebyshev",
    xlabel="x", ylabel="p(x)",
    legendfontsize=7
)

plot!(
    p2, x_interp, y_cheb_int,
    lw=2, color=:blue,
    label="Interpolador, d=$d, erro=$err_cheb_int"
)

scatter!(
    p2, X_cheb_int, b_cheb_int,
    ms=3, color=:black,
    label="nós de Chebyshev"
)

push!(plts, p2)


# ============================================================
# 3) EXTRAPOLACAO: AJUSTE EM [0,1], AVALIACAO EM [-1,0]
#    COM PONTOS UNIFORMES
# ============================================================

X_uni_ext = collect(range(0.0, 1.0, length=n))
b_uni_ext = f_test.(X_uni_ext)

V_uni_ext = vandermonde_matrix(X_uni_ext, d)
coef_uni_ext = V_uni_ext \ b_uni_ext

y_fit_uni_ext = vandermonde_matrix(x_fit_ext, d) * coef_uni_ext
y_out_uni_ext = vandermonde_matrix(x_out_ext, d) * coef_uni_ext

y_true_fit_ext = f_test.(x_fit_ext)
y_true_out_ext = f_test.(x_out_ext)

err_fit_uni_ext = round(maximum(abs.(y_fit_uni_ext - y_true_fit_ext)), sigdigits=3)
err_out_uni_ext = round(maximum(abs.(y_out_uni_ext - y_true_out_ext)), sigdigits=3)

p3 = plot(
    x_all_ext, f_test.(x_all_ext),
    lw=1.5, ls=:dot, color=:gray,
    label="f verdadeira",
    xlims=(-1.05, 1.05), ylims=(-2.0, 2.0),
    title="Extrapolação: pontos uniformes",
    xlabel="x", ylabel="p(x)",
    legendfontsize=7
)

plot!(
    p3, x_fit_ext, y_fit_uni_ext,
    lw=2, color=:blue,
    label="ajuste em [0,1], erro=$err_fit_uni_ext"
)

plot!(
    p3, x_out_ext, y_out_uni_ext,
    lw=2, color=:red,
    label="extrap. em [-1,0], erro=$err_out_uni_ext"
)

scatter!(
    p3, X_uni_ext, b_uni_ext,
    ms=3, color=:black,
    label="pontos de ajuste"
)

vline!(
    p3, [0.0],
    lw=1, ls=:dash, color=:black,
    label=""
)

push!(plts, p3)


# ============================================================
# 4) EXTRAPOLACAO: AJUSTE EM [0,1], AVALIACAO EM [-1,0]
#    COM NOS DE CHEBYSHEV
# ============================================================

X_cheb_ext = chebyshev_nodes(0.0, 1.0, n)
b_cheb_ext = f_test.(X_cheb_ext)

V_cheb_ext = vandermonde_matrix(X_cheb_ext, d)
coef_cheb_ext = V_cheb_ext \ b_cheb_ext

y_fit_cheb_ext = vandermonde_matrix(x_fit_ext, d) * coef_cheb_ext
y_out_cheb_ext = vandermonde_matrix(x_out_ext, d) * coef_cheb_ext

err_fit_cheb_ext = round(maximum(abs.(y_fit_cheb_ext - y_true_fit_ext)), sigdigits=3)
err_out_cheb_ext = round(maximum(abs.(y_out_cheb_ext - y_true_out_ext)), sigdigits=3)

p4 = plot(
    x_all_ext, f_test.(x_all_ext),
    lw=1.5, ls=:dot, color=:gray,
    label="f verdadeira",
    xlims=(-1.05, 1.05), ylims=(-2.0, 2.0),
    title="Extrapolação: nós de Chebyshev",
    xlabel="x", ylabel="p(x)",
    legendfontsize=7
)

plot!(
    p4, x_fit_ext, y_fit_cheb_ext,
    lw=2, color=:blue,
    label="ajuste em [0,1], erro=$err_fit_cheb_ext"
)

plot!(
    p4, x_out_ext, y_out_cheb_ext,
    lw=2, color=:red,
    label="extrap. em [-1,0], erro=$err_out_cheb_ext"
)

scatter!(
    p4, X_cheb_ext, b_cheb_ext,
    ms=3, color=:black,
    label="nós de Chebyshev"
)

vline!(
    p4, [0.0],
    lw=1, ls=:dash, color=:black,
    label=""
)

push!(plts, p4)


# ============================================================
# FIGURA FINAL
# ============================================================

plot(
    plts...,
    layout=(2,2),
    size=(1200, 760),
    margin=5Plots.mm,
    plot_title="Interpolação e extrapolação com pontos uniformes e nós de Chebyshev"
)
```

```julia
# Comparacao com grau alto:
#
# Mesmos 4 graficos:
# 1) Interpolacao em [-1,1] com pontos uniformes
# 2) Interpolacao em [-1,1] com nos de Chebyshev
# 3) Extrapolacao: ajuste em [0,1], avaliacao em [-1,0], pontos uniformes
# 4) Extrapolacao: ajuste em [0,1], avaliacao em [-1,0], nos de Chebyshev
#
# Em cada grafico comparamos:
# - Base canonica / Vandermonde
# - Base de Lagrange baricentrica

# Funcao-teste
f_test(x) = 1 / (1 + 25x^2)

# Mapeia os nos de Chebyshev de [-1,1] para [a,b]
function chebyshev_nodes_interval(a, b, m)
    z = chebyshev_nodes(m)
    x = (a + b)/2 .+ (b - a)/2 .* z
    return sort(x)
end

# Parametros
d = 40
n = d + 1

# Malhas de avaliacao
x_interp = collect(range(-1.0, 1.0, length=1000))

x_fit_ext = collect(range(0.0, 1.0, length=700))
x_out_ext = collect(range(-1.0, 0.0, length=700))
x_all_ext = collect(range(-1.0, 1.0, length=1400))

plts_comp = []

# ============================================================
# Funcao auxiliar para limitar visualmente valores muito grandes
# ============================================================

function safe_ylim(vals...; ymin=-5.0, ymax=5.0)
    all_vals = reduce(vcat, [collect(v) for v in vals])
    finite_vals = all_vals[isfinite.(all_vals)]

    if length(finite_vals) == 0
        return (ymin, ymax)
    end

    M = maximum(abs.(finite_vals))
    M = min(max(M, 1.5), ymax)

    return (-M, M)
end


# ============================================================
# 1) INTERPOLACAO EM [-1,1] COM PONTOS UNIFORMES
# ============================================================

X_uni_int = collect(range(-1.0, 1.0, length=n))
b_uni_int = f_test.(X_uni_int)

# Base canonica
V_uni_int = vandermonde_matrix(X_uni_int, d)
coef_uni_int = V_uni_int \ b_uni_int

y_vand_uni_int = vandermonde_matrix(x_interp, d) * coef_uni_int

# Base de Lagrange
L_uni_int = lagrange_barycentric(x_interp, d; nodes=X_uni_int)
y_lag_uni_int = L_uni_int * b_uni_int

# Funcao verdadeira
y_true_int = f_test.(x_interp)

err_vand_uni_int = round(maximum(abs.(y_vand_uni_int - y_true_int)), sigdigits=3)
err_lag_uni_int  = round(maximum(abs.(y_lag_uni_int  - y_true_int)), sigdigits=3)
diff_uni_int     = round(maximum(abs.(y_vand_uni_int - y_lag_uni_int)), sigdigits=3)

yl = safe_ylim(y_true_int, y_vand_uni_int, y_lag_uni_int; ymax=5.0)

p1 = plot(
    x_interp, y_true_int,
    lw=1.5, ls=:dot, color=:gray,
    label="f verdadeira",
    xlims=(-1.05, 1.05), ylims=yl,
    title="Interpolação: pontos uniformes",
    xlabel="x", ylabel="p(x)",
    legendfontsize=7
)

plot!(
    p1, x_interp, y_vand_uni_int,
    lw=2, color=:blue,
    label="Vandermonde, erro=$err_vand_uni_int"
)

plot!(
    p1, x_interp, y_lag_uni_int,
    lw=2, color=:orange,
    label="Lagrange, erro=$err_lag_uni_int"
)

scatter!(
    p1, X_uni_int, b_uni_int,
    ms=2.5, color=:black,
    label="pontos"
)

push!(plts_comp, p1)


# ============================================================
# 2) INTERPOLACAO EM [-1,1] COM NOS DE CHEBYSHEV
# ============================================================

X_cheb_int = chebyshev_nodes_interval(-1.0, 1.0, n)
b_cheb_int = f_test.(X_cheb_int)

# Base canonica
V_cheb_int = vandermonde_matrix(X_cheb_int, d)
coef_cheb_int = V_cheb_int \ b_cheb_int

y_vand_cheb_int = vandermonde_matrix(x_interp, d) * coef_cheb_int

# Base de Lagrange
L_cheb_int = lagrange_barycentric(x_interp, d; nodes=X_cheb_int)
y_lag_cheb_int = L_cheb_int * b_cheb_int

err_vand_cheb_int = round(maximum(abs.(y_vand_cheb_int - y_true_int)), sigdigits=3)
err_lag_cheb_int  = round(maximum(abs.(y_lag_cheb_int  - y_true_int)), sigdigits=3)
diff_cheb_int     = round(maximum(abs.(y_vand_cheb_int - y_lag_cheb_int)), sigdigits=3)

yl = safe_ylim(y_true_int, y_vand_cheb_int, y_lag_cheb_int; ymax=5.0)

p2 = plot(
    x_interp, y_true_int,
    lw=1.5, ls=:dot, color=:gray,
    label="f verdadeira",
    xlims=(-1.05, 1.05), ylims=yl,
    title="Interpolação: nós de Chebyshev",
    xlabel="x", ylabel="p(x)",
    legendfontsize=7
)

plot!(
    p2, x_interp, y_vand_cheb_int,
    lw=2, color=:blue,
    label="Vandermonde, erro=$err_vand_cheb_int"
)

plot!(
    p2, x_interp, y_lag_cheb_int,
    lw=2, color=:orange,
    label="Lagrange, erro=$err_lag_cheb_int"
)

scatter!(
    p2, X_cheb_int, b_cheb_int,
    ms=2.5, color=:black,
    label="nós"
)

push!(plts_comp, p2)


# ============================================================
# 3) EXTRAPOLACAO COM PONTOS UNIFORMES
#    AJUSTE EM [0,1], AVALIACAO EM [-1,0]
# ============================================================

X_uni_ext = collect(range(0.0, 1.0, length=n))
b_uni_ext = f_test.(X_uni_ext)

# Base canonica
V_uni_ext = vandermonde_matrix(X_uni_ext, d)
coef_uni_ext = V_uni_ext \ b_uni_ext

y_fit_vand_uni_ext = vandermonde_matrix(x_fit_ext, d) * coef_uni_ext
y_out_vand_uni_ext = vandermonde_matrix(x_out_ext, d) * coef_uni_ext

# Base de Lagrange
L_fit_uni_ext = lagrange_barycentric(x_fit_ext, d; nodes=X_uni_ext)
L_out_uni_ext = lagrange_barycentric(x_out_ext, d; nodes=X_uni_ext)

y_fit_lag_uni_ext = L_fit_uni_ext * b_uni_ext
y_out_lag_uni_ext = L_out_uni_ext * b_uni_ext

# Verdadeiro
y_true_fit_ext = f_test.(x_fit_ext)
y_true_out_ext = f_test.(x_out_ext)

err_out_vand_uni_ext = round(maximum(abs.(y_out_vand_uni_ext - y_true_out_ext)), sigdigits=3)
err_out_lag_uni_ext  = round(maximum(abs.(y_out_lag_uni_ext  - y_true_out_ext)), sigdigits=3)

diff_uni_ext = round(
    maximum(abs.(vcat(y_fit_vand_uni_ext, y_out_vand_uni_ext) .-
                 vcat(y_fit_lag_uni_ext,  y_out_lag_uni_ext))),
    sigdigits=3
)

yl = safe_ylim(
    f_test.(x_all_ext),
    y_fit_vand_uni_ext,
    y_out_vand_uni_ext,
    y_fit_lag_uni_ext,
    y_out_lag_uni_ext;
    ymax=5.0
)

p3 = plot(
    x_all_ext, f_test.(x_all_ext),
    lw=1.5, ls=:dot, color=:gray,
    label="f verdadeira",
    xlims=(-1.05, 1.05), ylims=yl,
    title="Extrapolação: pontos uniformes",
    xlabel="x", ylabel="p(x)",
    legendfontsize=7
)

# Vandermonde
plot!(
    p3, x_fit_ext, y_fit_vand_uni_ext,
    lw=2, color=:blue,
    label="Vand. em [0,1]"
)

plot!(
    p3, x_out_ext, y_out_vand_uni_ext,
    lw=2, color=:red,
    label="Vand. extrap., erro=$err_out_vand_uni_ext"
)

# Lagrange
plot!(
    p3, x_fit_ext, y_fit_lag_uni_ext,
    lw=2, color=:orange,
    ls=:dash,
    label="Lag. em [0,1]"
)

plot!(
    p3, x_out_ext, y_out_lag_uni_ext,
    lw=2, color=:purple,
    ls=:dash,
    label="Lag. extrap., erro=$err_out_lag_uni_ext"
)

scatter!(
    p3, X_uni_ext, b_uni_ext,
    ms=2.5, color=:black,
    label="pontos"
)

vline!(
    p3, [0.0],
    lw=1, ls=:dash, color=:black,
    label=""
)

push!(plts_comp, p3)


# ============================================================
# 4) EXTRAPOLACAO COM NOS DE CHEBYSHEV
#    AJUSTE EM [0,1], AVALIACAO EM [-1,0]
# ============================================================

X_cheb_ext = chebyshev_nodes_interval(0.0, 1.0, n)
b_cheb_ext = f_test.(X_cheb_ext)

# Base canonica
V_cheb_ext = vandermonde_matrix(X_cheb_ext, d)
coef_cheb_ext = V_cheb_ext \ b_cheb_ext

y_fit_vand_cheb_ext = vandermonde_matrix(x_fit_ext, d) * coef_cheb_ext
y_out_vand_cheb_ext = vandermonde_matrix(x_out_ext, d) * coef_cheb_ext

# Base de Lagrange
L_fit_cheb_ext = lagrange_barycentric(x_fit_ext, d; nodes=X_cheb_ext)
L_out_cheb_ext = lagrange_barycentric(x_out_ext, d; nodes=X_cheb_ext)

y_fit_lag_cheb_ext = L_fit_cheb_ext * b_cheb_ext
y_out_lag_cheb_ext = L_out_cheb_ext * b_cheb_ext

err_out_vand_cheb_ext = round(maximum(abs.(y_out_vand_cheb_ext - y_true_out_ext)), sigdigits=3)
err_out_lag_cheb_ext  = round(maximum(abs.(y_out_lag_cheb_ext  - y_true_out_ext)), sigdigits=3)

diff_cheb_ext = round(
    maximum(abs.(vcat(y_fit_vand_cheb_ext, y_out_vand_cheb_ext) .-
                 vcat(y_fit_lag_cheb_ext,  y_out_lag_cheb_ext))),
    sigdigits=3
)

yl = safe_ylim(
    f_test.(x_all_ext),
    y_fit_vand_cheb_ext,
    y_out_vand_cheb_ext,
    y_fit_lag_cheb_ext,
    y_out_lag_cheb_ext;
    ymax=5.0
)

p4 = plot(
    x_all_ext, f_test.(x_all_ext),
    lw=1.5, ls=:dot, color=:gray,
    label="f verdadeira",
    xlims=(-1.05, 1.05), ylims=yl,
    title="Extrapolação: nós de Chebyshev",
    xlabel="x", ylabel="p(x)",
    legendfontsize=7
)

# Vandermonde
plot!(
    p4, x_fit_ext, y_fit_vand_cheb_ext,
    lw=2, color=:blue,
    label="Vand. em [0,1]"
)

plot!(
    p4, x_out_ext, y_out_vand_cheb_ext,
    lw=2, color=:red,
    label="Vand. extrap., erro=$err_out_vand_cheb_ext"
)

# Lagrange
plot!(
    p4, x_fit_ext, y_fit_lag_cheb_ext,
    lw=2, color=:orange,
    ls=:dash,
    label="Lag. em [0,1]"
)

plot!(
    p4, x_out_ext, y_out_lag_cheb_ext,
    lw=2, color=:purple,
    ls=:dash,
    label="Lag. extrap., erro=$err_out_lag_cheb_ext"
)

scatter!(
    p4, X_cheb_ext, b_cheb_ext,
    ms=2.5, color=:black,
    label="nós"
)

vline!(
    p4, [0.0],
    lw=1, ls=:dash, color=:black,
    label=""
)

push!(plts_comp, p4)


# ============================================================
# FIGURA FINAL
# ============================================================

plot(
    plts_comp...,
    layout=(2,2),
    size=(1250, 800),
    margin=5Plots.mm,
    plot_title="Grau alto: base canônica vs base de Lagrange baricêntrica, d=$d"
)
```

#### Comentários sobre os gráficos acima

**Condicionamento (gráfico de barras):** A Extrapolação 2 apresenta os piores condicionamentos de todos os cenários estudados. O problema é estrutural: os intervalos $[0,1]$ e $[-1,0]$ são completamente separados, então o polinômio precisa ser extrapolado por uma distância comparável ao comprimento do intervalo de ajuste. Isso produz oscilações de Runge máximas.

Curiosamente, Chebyshev em $[0,1]$ ajuda pouco aqui (curva verde quase idêntica à vermelha para $d$ médio), confirmando que a instabilidade não vem da escolha dos nós mas sim da geometria do problema: nenhuma distribuição de pontos em $[0,1]$ pode controlar o comportamento de um polinômio em $[-1,0]$.

**Visualização direta (gráfico de curvas):** O efeito é imediato. Para $d = 3$, o polinômio permanece razoável em ambos os lados. Para $d = 8$, as avaliações em $[-1, 0]$ (vermelho) já mostram desvios notáveis da função verdadeira (pontilhado cinza). Para $d = 15$, o polinômio explode completamente fora do domínio de ajuste, atingindo valores de magnitude $\sim 10$ mesmo que a função original oscile apenas entre $-1$ e $1$.

**Adianta usar Chebyshev?** Para a Extrapolação 2, a resposta é **essencialmente não**. A razão é que os pontos de Chebyshev minimizam o polinômio de Lebesgue em $[0,1]$, mas esse controle não se estende a $[-1,0]$. Qualquer polinômio de grau alto avaliado fora do seu intervalo de ajuste crescerá sem controle — a escolha dos nós pode mitigar marginalmente o efeito, mas não pode resolver um problema que é fundamentalmente geométrico: estamos pedindo ao polinômio informações sobre uma região onde não possui dados.
